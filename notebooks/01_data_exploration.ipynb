# 01. Data Exploration & Time-Window Analysis
**Project:** Identifying Emerging Topics Before They Become Visible (ML-T2-008)

In this notebook, we:
1. Load the cleaned Reddit dataset (`reddit_cleaned_sample.csv`).
2. Inspect the distribution of posts over chronological time windows ($W_0, W_1, \dots, W_5$).
3. Check word counts and vocabulary.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/reddit_cleaned_sample.csv')
print(f"Total posts: {len(df)}")
df.head()
```

```python
# Posts per time window
window_summary = df.groupby('window_id').agg(
    start_time=('datetime_utc', 'min'),
    end_time=('datetime_utc', 'max'),
    post_count=('id', 'count')
).reset_index()

window_summary
```
